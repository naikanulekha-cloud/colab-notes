```python
""" continuous distribution
 packet delivery time has mean 40 minutes and sd 5 minutes
 find proportion under 47 minutes"""

from scipy.stats import norm

mu, sigma, x = 40, 5, 47

z = (x - mu) / sigma
p = norm.cdf(z)

print("z-score:", round(z, 1))
print("p(time<=47):", round(p, 4))
```

```python
""" a companey claims that the average time requiredd to packaging is 30  min a sample of 10 orders has the following ppacking time are 28,31,29,35,30,27,32,29,34 and 30minn
# at 5% los
# use 1 sample t test to  determine weeather
# the actual average packing time is signifintly diff from the companneys claim avarage"""
from scipy.stats import ttest_1samp
import numpy as np

data = [28, 31, 29, 35, 30, 27, 32, 29, 34, 30]

mu = 30

mean = np.mean(data)
t_test, p_value = ttest_1samp(data, mu)

print("mean:", round(mean, 1))
print("t test:", round(t_test, 2))
print("p value:", round(p_value, 4))
```



1. Chi-Square Test: Used to compare observed and expected frequencies.

2. Observed Frequency: Actual collected values.

3. Expected Frequency: Predicted values.

4. Null Hypothesis (H₀): No significant difference.

5. Alternative Hypothesis (H₁): Significant difference exists.

6. P-value: Helps decide whether to accept or reject H₀.

7. Significance Level: Usually 0.05 (5%)
